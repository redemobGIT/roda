```python
import os
from pathlib import Path

import basedosdados as bd
from dotenv import load_dotenv
import geobr
import geopandas as gpd
import geobr
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
load_dotenv()
GCLOUD_ID = os.getenv("GCLOUD_ID")


OUT_DIR = Path("../outputs")
OUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
INPATH = Path("../database/grade_id26_2022.zip")

grid = gpd.read_file(INPATH).to_crs(31983)
```

```python
rmrj = geobr.read_metro_area().query("name_metro == 'RM Rio de Janeiro'").to_crs(31983)
rmrj = rmrj.loc[rmrj.name_muni.isin(["Niterói", 'São Gonçalo', 'Itaboraí'])]
rmrj = rmrj.union_all().convex_hull
```

```python
grid = grid.loc[grid.intersects(rmrj)]
grid = grid.loc[grid.TOTAL >= 1]
```

```python
grid.to_file(OUT_DIR / "grid_rmrj.gpkg")
```
